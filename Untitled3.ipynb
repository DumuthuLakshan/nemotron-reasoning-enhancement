```python
from transformers import pipeline

pipe = pipeline("text-generation", model=model, tokenizer=tokenizer)

def generate_answer(question):
    prompt = f"""Solve the problem carefully.
Output only the final answer.
Do not add extra explanation.

Question: {question}

Answer:
"""
    out = pipe(
        prompt,
        max_new_tokens=20,
        do_sample=False,
        return_full_text=True
    )[0]["generated_text"]
    return out
```

```python
import re

def extract_prediction(text):
    if not isinstance(text, str):
        return ""
    lines = [line.strip() for line in text.splitlines() if line.strip()]
    if lines:
        return lines[-1]
    return ""
```

```python
results = []
for _, row in test_df.iterrows():
    out = generate_answer(row["prompt"])
    pred = extract_prediction(out)
    results.append({"id": row["id"], "prediction": pred})

submission = pd.DataFrame(results)
submission.to_csv("/content/submission.csv", index=False)
print(submission.head())
```

```python
import pandas as pd

test_df = pd.read_csv("test.csv")

print(test_df.head())
print("Total test samples:", len(test_df))
```

```python
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)

def generate_answer(question):
    prompt = f"""Solve the problem step by step.
Give final answer in \\boxed{{}}.

Question: {question}
Answer:
"""

    output = pipe(prompt, max_new_tokens=150)[0]["generated_text"]
    return output
```

```python
results = []

for _, row in test_df.iterrows():
    output = generate_answer(row["prompt"])

    results.append({
        "prompt": row["prompt"],
        "model_output": output
    })

results_df = pd.DataFrame(results)

print(results_df.head())
```

```python
print(results_df["model_output"][0])
```

```python
import re

def extract_boxed(text):
    match = re.search(r"\\boxed\{(.*?)\}", text)
    return match.group(1) if match else None

results_df["predicted_answer"] = results_df["model_output"].apply(extract_boxed)
```

```python
correct = 0

for i, row in test_df.iterrows():
    if "answer" in test_df.columns:
        if str(results_df["predicted_answer"][i]) == str(row["answer"]):
            correct += 1

accuracy = correct / len(test_df)
print("Accuracy:", accuracy)
```

```python
submission = pd.DataFrame({
    "id": test_df["id"],
    "prediction": results_df["predicted_answer"]
})

submission.to_csv("/content/submission.csv", index=False)
```

```python
import re

def extract_prediction(text):
    if not isinstance(text, str):
        return ""

    # 1. Try boxed answer first
    match = re.search(r"\\boxed\{(.*?)\}", text)
    if match:
        return match.group(1).strip()

    # 2. Try "Final answer: ..."
    match = re.search(r"Final answer:\s*(.+)", text, re.IGNORECASE)
    if match:
        return match.group(1).strip()

    # 3. Try last non-empty line
    lines = [line.strip() for line in text.splitlines() if line.strip()]
    if lines:
        return lines[-1]

    # 4. Otherwise return empty string
    return ""
```
